## This notebook demonstrates how to execute feature extraction, SLAT encoding, and SLAT decoding

```python
import os
from feature_extractor import FeatureExtractor
from latent_encoder import LatentEncoder
from slat_decoder import SLATDecoder
```

### Setup everything for SLAT generation

- n_views is recommended to be in range(120, 200)
- batch_size== 150 takes over 16G of VRAM

```python
extractor = FeatureExtractor(batch_size=150, n_views=150)
slat_encoder = LatentEncoder()
```

If **ref_mesh_path** is given, the input mesh will be normalized and shifted to match the normalization settings of this reference mesh.

```python
example_name = "car-luggage-rack"  # TODO: Pick an example from assets/example_objects
root_dir = os.path.dirname(os.path.dirname(os.getcwd()))
print(root_dir)
input_mesh_path = os.path.join(root_dir, "assets", "example_objects", example_name, "unmodified.obj")
ref_mesh_path = os.path.join(root_dir, "assets", "example_objects", example_name, "original.obj")
```

1. Extract features from multiple rendered views
2. Aggregate these features into a **Structured LATent**

```python
feat_path = extractor.run_extractor(input_mesh_path, ref_mesh_path, force_render=True)
slat_encoder = LatentEncoder()
slat_path = slat_encoder.run_slat_encoder(feat_path)
print("Done! SLAT path:", slat_path)
```

Check the SLAT's size

```python
import numpy as np

slat_path = ""  # TODO: Enter the path to the .npz file
data = np.load(slat_path)
print(type(data))
print(np.shape(data['coords']))
print(np.shape(data['feats'])) 
```

### Decode the SLAT into a mesh

```python
output_dir = ""  # TODO: Enter the path where the output mesh is going to be saved.
slat_decoder = SLATDecoder()
slat_decoder.create_output_folder(output_dir)
slat_decoder.decode(slat_path)
print("Decoding completed.")
```
